# Compare csvs

```python
# --- Calea către CSV-uri ---
csv1_path = "features_timeout_15s_changed_header_filled.csv"
csv2_path = "features_timeout_15s_nonvpn_filtered.csv"

# --- Încarcă CSV-urile ---
df1 = pd.read_csv(csv1_path)
df2 = pd.read_csv(csv2_path)

# --- Găsește coloanele comune ---
common_columns = df1.columns.intersection(df2.columns)
common_columns = common_columns.drop(labels=["Label"])  # Exclude coloana 'label' dacă există
print(f"Coloane comune ({len(common_columns)}): {list(common_columns)}\n")

# --- Funcție pentru statistici ---
def compute_stats(df, features):
    summary = pd.DataFrame({
        "min": df[features].min(),
        "max": df[features].max(),
        "mean": df[features].mean(),
        "median": df[features].median(),
        "std": df[features].std(),
        "nan_count": df[features].isna().sum(),
        "zero_count": (df[features] == 0).sum(),
        "inf_count": np.isinf(df[features]).sum()
    })
    return summary

# --- Statistici pentru fiecare CSV ---
print("=== Statistici CSV 1 ===")
stats1 = compute_stats(df1, common_columns)
print(stats1)

print("\n=== Statistici CSV 2 ===")
stats2 = compute_stats(df2, common_columns)
print(stats2)
```
